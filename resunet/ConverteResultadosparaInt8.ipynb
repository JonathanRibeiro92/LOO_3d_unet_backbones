```python
from skimage import io
import numpy as np
from matplotlib import pyplot as plt
import os
import gc
```

```python
from tifffile import imsave
import os

# name_file = '3D_result/'+ amostra + '_segmented.tif'
directory_path= '3D_result/'
# List all files and directories
files = [file for file in os.listdir(directory_path) if file.endswith('tif')]

# Print the files
for file in files:
    print(file)
```

```python
def readImageAndConvertInt8(file):
    directory_path = '3D_result/'
    fileimage = os.path.join(directory_path,file)
    image = io.imread(fileimage)
    image = image.astype(np.uint8)
    return image
```

```python
def generateInt8(amostra, image):
    directory = "3D_result/uint8/"

    # Check if the directory exists
    if not os.path.exists(directory):
        # Create the directory
        os.makedirs(directory)
    
    name_file = directory + amostra + '_segmented.tif'
    imsave(name_file, image)
```

```python
for file in files:
    amostra = file[:5]
    print(f'amostra {amostra} em processamento')
    image = readImageAndConvertInt8(file)
    generateInt8(amostra, image)
print(f'Todas amostras processadas')
```
